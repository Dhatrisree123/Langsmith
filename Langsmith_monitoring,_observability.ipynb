```python
# Let us configure gemini ai
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    temperature=0.2)

response = llm.invoke(
    "Explain LangSmith in simple terms.")
print(response.content)
```

```python
# Prompt Template
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        "You are an experienced data science trainer."
    ),
    (
        "human",
        "{question}"
    )
])
```

```python
# Chain
chain = prompt | llm
```

```python
response = chain.invoke({
    "question": "Explain K-Means clustering."})

print(response.content)
```

```python
# monitoring latency
import time

start_time = time.time()

response = chain.invoke({
    "question": "Explain neural networks."
})

end_time = time.time()

latency = end_time - start_time

print("Response:")
print(response.content)
print("\nLatency:", round(latency, 2), "seconds")
```

```python
# multiple requests
questions = [
    "What is supervised learning?",
    "What is unsupervised learning?",
    "What is reinforcement learning?",
    "What is deep learning?",
    "What is transfer learning?"
]

for question in questions:

    response = chain.invoke({
        "question": question
    })

    print("\n QUESTION:")
    print(question)

    print("\n ANSWER:")
    print(response.content)

    print("=" * 70)
```

```python
# Adding Metadata
response = chain.invoke(
    {
        "question": "Explain PCA."
    },
    config={
        "metadata": {
            "application": "data-science-tutor",
            "environment": "development",
            "user_type": "student"
        }
    }
)

print(response.content)
```

```python
# adding a tag
response = chain.invoke(
    {
        "question": "Explain classification."
    },
    config={
        "tags": [
            "data-science",
            "classification",
            "training"
        ]
    }
)

print(response.content)
```

# **AI Tutor Application**

```python
from langchain_core.prompts import ChatPromptTemplate
prompt = ChatPromptTemplate.from_messages([
    (
        "system",
        """
        You are an expert data science trainer.

        Explain concepts clearly.
        Use simple examples.
        Target audience: mid-career professionals.
        """
    ),
    (
        "human",
        "{question}"
    )
])

chain = prompt | llm
```

```python
# Let us create a function
def data_science_tutor(question):

    response = chain.invoke(
        {"question": question},
        config={
            "tags": ["data-science", "tutor"],
            "metadata": {
                "application": "data-science-tutor",
                "environment": "development"
            }
        }
    )

    return response.content
```

```python
answer = data_science_tutor(
    "Explain hierarchical clustering."
)

print(answer)
```
